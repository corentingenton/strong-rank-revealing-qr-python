```python
def randomized_qdp(M_sk=M_sk_kahan, M=M_kahan, l=1):
    Q_tilda, R, P, k = randomized_srrqr(M_sk=M_sk, M=M)
    print("Q_tilda", Q_tilda.shape)
    print("R", R.shape)
    print("P", P.shape)
    Q_bar, L_T = scipy.linalg.qr(R[:k+1, :].T, pivoting=False)
    print("L_T", L_T.shape)
    print("Q_bar", Q_bar.shape)
    U, D_l, V_T = scipy.sparse.linalg.svds(L_T.T, k=l)
    print(D_l.shape)
    print(D_l)
    D_l = D_l[::-1]
    print(D_l)
    U = U[:, ::-1]
    V_T = V_T[::-1, :]
    print("U", U.shape)
    print("D_l", D_l.shape)
    print("V_T", V_T.shape)
    L_l = U @ np.diag(D_l) @ V_T
    print("L_l", L_l.shape)
    m = M.shape[0] # What is the value of m ?
    print("Q_tilda m, k", Q_tilda[:m, :k+1].shape)
    Q = Q_tilda[:m+1, :k+1] @ U
    print("Q", Q.shape)
    P = P @ Q_bar @ V_T.T
    M_hat_l =  Q @ np.diag(D_l) @ P.T
    return Q, D_l, P, M_hat_l
```
